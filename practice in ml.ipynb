```python
x = data.drop("Spending", axis=1)
y = data["Spending"]
```

```python
from sklearn. model_selection import train_test_split
```

```python
xtrain, xtest, ytrain, ytest = train_test_split( x, y, test_size = 0.2, random_state = 42)
```

```python
from sklearn.linear_model import BayesianRidge
```

```python
model = BayesianRidge()
```

```python
model.fit(xtrain, ytrain)
```

```python
prediction, std = model.predict(xtest, return_std=True)
```

```python
prediction, std
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
print(mean_squared_error(ytest, prediction))
print(mean_absolute_error(ytest, prediction))
print(r2_score(ytest, prediction))
```

```python
import pandas as pd
```

```python
data2 = pd.read_excel(r"C:\Users\user\Desktop\Classification.xlsx")
data2
```

```python
x = data2.drop("Diabetes", axis=1)
y = data2["Diabetes"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size = 0.3, random_state = 42)
```

```python
from sklearn.naive_bayes import GaussianNB
```

```python
model = GaussianNB()
```

```python
model.fit(xtrain, ytrain)
```

```python
prediction = model.predict(xtest)
```

```python
prediction
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
print(accuracy_score(ytest, prediction))
print(confusion_matrix(ytest, prediction))
print(classification_report(ytest, prediction))
```

```python
import matplotlib.pyplot as plt
```

```python
residuals = ytest - prediction
```

```python
plt.figure(figsize=(8,5))
```

```python
plt.scatter(prediction, residuals)
plt.axhline(y=0, color='red', linestyle='--')
plt.title("Residuals plot")
plt.xlabel("Predicted value")
plt.ylabel("residuals")
plt.show()
```
