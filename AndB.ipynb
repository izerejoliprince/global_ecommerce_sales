```python
import random
```

```python
emailsA = 300
clicksA = 45

emailsB = 300
clicksB = 60
```

```python
rateA = clicksA / emailsA
rateB = clicksB / emailsB

print("Campaign A Conversion Rate:", round(rateA, 4))
print("Campaign B Conversion Rate:", round(rateB, 4))
```

```python
winsB = 0
```

```python
for i in range(1000):

    sampleA = random.betavariate(
        clicksA + 1,
        emailsA - clicksA + 1
    )

    sampleB = random.betavariate(
        clicksB + 1,
        emailsB - clicksB + 1
    )

    if sampleB > sampleA:
        winsB += 1
```

```python
probability = winsB / 1000

print("Probability Campaign B is Better:",
      round(probability,4))
```

```python
if probability > 0.90:
    print("Recommendation:")
    print("Choose Campaign B")
else:
    print("Recommendation:")
    print("Collect More Data")
```
