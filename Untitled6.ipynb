```python
import pandas as pd
```

```python
data = pd.read_excel(r"C:\Users\user\Desktop\Year1 and Year 2\year 2\2nd term\Business Forecasting\reviews.xlsx")
data
```

```python
x = data["Review"]
y = data["Sentiment"]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
vectorizer = CountVectorizer()
x_vector = vectorizer.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x_vector, y, test_size = 0.2, random_state=42)
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
model = MultinomialNB()
```

```python
model.fit(xtrain, ytrain)
```

```python
prediction = model.predict(xtest)
prediction
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
tree_model = DecisionTreeClassifier(random_state=42)
```

```python
tree_model.fit(xtrain, ytrain)
```

```python
tree_prediction = tree_model.predict(xtest)
tree_prediction
```
