```python
import pandas as pd
data = pd.read_excel(r"C:\Users\user\Desktop\spending.xlsx")
data
```

```python
x = data.drop("Spending", axis = 1)
y = data["Spending"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size = 0.3, random_state = 1)
```

```python
from sklearn.linear_model import BayesianRidge
```

```python
model = BayesianRidge()
```

```python
model.fit(xtrain, ytrain)
```

```python
prediction, std = model.predict(xtest, return_std=True)
```

```python
from sklearn.metrics import (mean_squared_error, mean_absolute_error, r2_score)
```

```python
print(mean_squared_error(ytest, prediction))
print(mean_absolute_error(ytest, prediction))
print(r2_score(ytest, prediction))
```

```python
import matplotlib.pyplot as plt
```

```python
residuals = ytest - prediction
```

```python
plt.figure(figsize=(8,5))
plt.scatter(prediction, residuals)
```
