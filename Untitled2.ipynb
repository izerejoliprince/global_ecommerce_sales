```python
import pandas as pd
data = pd.read_excel(r"C:\Users\user\Desktop\Year1 and Year 2\year 2\2nd term\Business Forecasting\regression.xlsx")
data
```

```python
x = data.drop("Spending", axis=1)
y = data["Spending"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size = 0.2, random_state=42)
```

```python
from sklearn. linear_model import BayesianRidge
```

```python
model = BayesianRidge()
```

```python
model.fit(xtrain, ytrain)
```

```python
prediction, std = model.predict(xtrain, return_std =True)
prediction, std
```

```python
from sklearn. metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
print(mean_squared_error(ytest, prediction))
print(mean_absolute_error(ytest, prediction))
print(r2_score(ytest, prediction))
```
