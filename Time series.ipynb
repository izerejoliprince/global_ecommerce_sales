```python
import numpy as np
```

```python
months = np.array([1,2,3,4,5,6])
sales = np.array([1000,1200,1300,1400,1450,1500])
```

```python
trend = np.polyfit(months, sales, 1)
```

```python
model = np.poly1d(trend)
```

```python
future_months = [7,8,9]
```

```python
forecast = model(future_months)
forecast
```
