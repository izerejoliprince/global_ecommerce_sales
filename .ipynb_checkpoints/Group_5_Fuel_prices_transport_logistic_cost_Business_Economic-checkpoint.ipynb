Problem Statement
</div>

Transportation and logistics costs are a major input cost for Rwandan businesses, particularly
those that depend on imported goods, road freight, or fuel-intensive operations. Because Rwanda
is landlocked, fuel price movements (driven largely by global oil markets, exchange rates, and
domestic taxation) pass through quickly into transport costs, which in turn affect the prices of
almost every good and service in the economy.

This project investigates: **How do changes in fuel (gasoline) prices affect transportation costs
in Rwanda, and what does this mean for business decision-making?**

We use the **Transport component of the Consumer Price Index (CPI)** as a proxy for transportation/
logistics costs, since Rwanda does not publish a single dedicated "logistics cost index." Movements
in the CPI-Transport sub-index reflect changes in the prices businesses and consumers pay for
transport services (bus fares, freight, vehicle maintenance, fuel purchased by households, etc.).

## 1. IMPORT LIBRARIES

```python
# --- 1. SETUP & IMPORTS ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error
import statsmodels.api as sm

sns.set_style('whitegrid')
plt.rcParams['figure.dpi'] = 120
plt.rcParams['font.size'] = 11
pd.set_option('display.max_columns', None)
```

## 2. Data Sources

| Variable | Period | Source | Notes |
|---|---|---|---|
| `Gasoline_USD_per_L` | 2004–2025 | RURA pump-price bulletins (2009–2025) & World Bank/GIZ International Fuel Prices indicator `EP.PMP.SGAS.CD` (2004–2008) | 2005 & 2007 are linear interpolations, flagged in `Data_Type` |
| `CPI_Transport_Annual` | 2009–2025 | NISR Consumer Price Index survey, Transport division sub-index (2014=100 base) | NISR's rebased CPI series begins Feb-2009, so no earlier annual figure exists at this methodology |

Because the two series have different available lengths, the notebook keeps **two working frames**:
* `df_full` : 22 years (2004–2025) for the fuel-price-only trend/forecast work
* `df_cc`  :  17-year **complete-case** frame (2009–2025) used for every analysis that needs *both* variables (correlation, regression, scenario modelling)


```python
# --- 3. LOAD RAW DATA ---
df_raw = pd.read_csv('raw_fuel_cpi_rwanda_2004_2025.csv')
df_raw
```

```python
# --- 4. DATA CLEANING ---
# Check types, missing values, duplicates
print(df_raw.dtypes)
print("\nMissing values:\n", df_raw.isna().sum())
print("\nDuplicate years:", df_raw['YEAR'].duplicated().sum())

df_full = df_raw.copy()
df_full['Fuel_YoY_pct'] = df_full['Gasoline_USD_per_L'].pct_change() * 100
df_full['CPI_YoY_pct']  = df_full['CPI_Transport_Annual'].pct_change() * 100
df_full.to_csv('cleaned_fuel_cpi_rwanda.csv', index=False)

# Complete-case frame (both variables present) = 2009-2025
df_cc = df_full.dropna(subset=['CPI_Transport_Annual']).reset_index(drop=True)
df_cc.to_csv('cleaned_fuel_cpi_2009_2025_complete.csv', index=False)

print(f"\nFull fuel-price series: {df_full.YEAR.min()}-{df_full.YEAR.max()}  (n={len(df_full)})")
print(f"Complete-case series (fuel + CPI): {df_cc.YEAR.min()}-{df_cc.YEAR.max()}  (n={len(df_cc)})")
df_cc.head()
```

## 5. Basic Data Analysis :  Descriptive Statistics

We summarize the central tendency and spread of both variables using `.describe()`, and
report the mean, median, and standard deviation explicitly.

```python
desc = df_cc[['Gasoline_USD_per_L','CPI_Transport_Annual']].describe()
desc.loc['skew'] = df_cc[['Gasoline_USD_per_L','CPI_Transport_Annual']].skew()
desc
```

```python
for col in ["Gasoline_USD_per_L", "CPI_Transport_Annual"]:
    print(f"--- {col} ---")
    print(f"Mean:   {df_cc[col].mean():.3f}")
    print(f"Median: {df_cc[col].median():.3f}")
    print(f"Std Dev:{df_cc[col].std():.3f}")
    print(f"Min:    {df_cc[col].min():.3f}  (Year {df_cc.loc[df_cc[col].idxmin(),'YEAR']})")
    print(f"Max:    {df_cc[col].max():.3f}  (Year {df_cc.loc[df_cc[col].idxmax(),'YEAR']})")
    print()
```

**Interpretation:** Over 2009–2025, Rwanda's pump gasoline price averaged **USD 1.34/L**
(std dev 0.18), ranging from a low of USD 1.17/L (2016) to a high of USD 1.73/L (2012).
The Transport CPI averaged **116.6** and more than doubled in dispersion relative to its mean growth,
reflecting a structural upward trend rather than a stable price level , consistent with Rwanda's
rising transport-fare and vehicle-cost environment (aggravated by the March 2024 removal of public
transport fare subsidies).  

Over all, the sample period, gasoline prices in Rwanda have shown moderate
variation (driven by global oil price swings, e.g. the 2014-2016 oil price decline and the
2021-2022 post-pandemic price spike), while the CPI-Transport index has risen fairly
steadily, reflecting persistent cost-push inflation in transport services even in years when
fuel prices were flat or falling.

## 4. COMPARIZON BETWEEN FUEL PRICES AND CPI FOR TRANSPORT

Two distinct regimes are visible in the fuel-price series:
1. **2004–2012:** steady rise (global oil boom, growing Rwandan import demand) up to a peak of USD 1.73/L.
2. **2012–2016:** decline (global oil price collapse of 2014–2016).
3. **2017–2021:** flat/administratively fixed price (RURA price-stabilisation policy kept pump prices
   almost static at ≈USD 1.17/L for five straight years).
4. **2022–2025:** renewed sharp rise (global energy-price shock following Russia's invasion of Ukraine,
   plus Rwandan Franc depreciation against the US dollar).

The Transport CPI shows a much smoother, near-monotonic **upward trend** throughout 2009–2025, with a
visible acceleration from 2022 onward and a sharp jump in 2024 coinciding with the government's removal
of public-transport fare subsidies (reported by NISR/allAfrica, April 2024).


```python
# --- Graph 3: Dual-axis comparison ---
fig, ax1 = plt.subplots(figsize=(9,5))
ax2 = ax1.twinx()
ax1.plot(df_cc.YEAR, df_cc.Gasoline_USD_per_L, color='#1f4e79', marker='o', label='Gasoline USD/L')
ax2.plot(df_cc.YEAR, df_cc.CPI_Transport_Annual, color='#8b1e3f', marker='s', label='CPI Transport')
ax1.set_xlabel('Year'); ax1.set_ylabel('Gasoline USD/L', color='#1f4e79')
ax2.set_ylabel('CPI Transport Index', color='#8b1e3f')
plt.title('Fuel Price vs Transport CPI, Rwanda 2009-2025', fontweight='bold')
l1,lb1 = ax1.get_legend_handles_labels(); l2,lb2 = ax2.get_legend_handles_labels()
ax1.legend(l1+l2, lb1+lb2, loc='upper left')
plt.tight_layout(); plt.savefig('fig3_dual_axis.png'); plt.show()
```

## 7. CORRELATION ANALYSIS

```python
# Level correlation
r_level, p_level = stats.pearsonr(df_cc.Gasoline_USD_per_L, df_cc.CPI_Transport_Annual)
print(f"Level correlation:      r = {r_level:.3f},  p = {p_level:.4f}")

# Year-on-year %% change correlation (removes spurious trend-driven correlation)
sub = df_cc.dropna(subset=['Fuel_YoY_pct','CPI_YoY_pct'])
r_yoy, p_yoy = stats.pearsonr(sub.Fuel_YoY_pct, sub.CPI_YoY_pct)
print(f"YoY %% change correlation: r = {r_yoy:.3f},  p = {p_yoy:.4f}")
```

```python
fig, axes = plt.subplots(1,2, figsize=(13,5.5))
sns.regplot(x='Gasoline_USD_per_L', y='CPI_Transport_Annual', data=df_cc, ax=axes[0],
            scatter_kws={'color':'#1f4e79','s':55}, line_kws={'color':'#e07b39'})
axes[0].set_title(f'Level correlation (r={r_level:.2f}, n.s.)', fontweight='bold')

sns.regplot(x='Fuel_YoY_pct', y='CPI_YoY_pct', data=sub, ax=axes[1],
            scatter_kws={'color':'#1f4e79','s':55}, line_kws={'color':'#e07b39'})
axes[1].axhline(0,color='grey',lw=0.8); axes[1].axvline(0,color='grey',lw=0.8)
axes[1].set_title(f'YoY %% change correlation (r={r_yoy:.2f}, p={p_yoy:.3f})', fontweight='bold')
plt.tight_layout(); plt.savefig('fig4_correlation_scatter.png'); plt.show()
```

**Interpretation:** The correlation between the *level* of the fuel price and the *level* of the
Transport CPI is essentially zero (r ≈ 0.01, not significant). This is not a data error, it reflects
a real structural feature of Rwanda's fuel market: RURA administratively sets and periodically revises
pump-price caps and margins, so retail fuel prices do not move continuously with world oil prices, while
the Transport CPI is driven by many other forces (bus/moto fare regulation, vehicle import costs, road
tolls, the 2024 subsidy removal). Both series also carry their own long-run trends, which alone can
create a misleading level-correlation.

Differencing the data into **year-on-year percentage changes** :  the econometrically correct way to
test co-movement between two trending series , shows a **moderate positive relationship (r ≈ 0.48,
p ≈ 0.06, borderline-significant at 10%)**: years in which pump prices rise faster tend to be years in
which transport costs also rise faster, i.e. **partial, delayed pass-through** rather than an automatic
one-for-one relationship. This is the more economically meaningful result and the one used below for
predictive modelling.

## 8. Predictive Modelling : Simple Linear Regression

```python
# Predict CPI Transport level from fuel price level (scikit-learn)
X = df_cc[['Gasoline_USD_per_L']].values
y = df_cc['CPI_Transport_Annual'].values

lr = LinearRegression().fit(X, y)
pred = lr.predict(X)
print(f"CPI_Transport = {lr.intercept_:.2f} + {lr.coef_[0]:.2f} * Fuel_price")
print(f"R-squared = {r2_score(y, pred):.4f}   MAE = {mean_absolute_error(y, pred):.2f}")

# Full statistical inference table (statsmodels)
Xc = sm.add_constant(df_cc['Gasoline_USD_per_L'])
model = sm.OLS(df_cc['CPI_Transport_Annual'], Xc).fit()
print(model.summary())
```

**Interpretation:** The regression confirms the correlation result ,  fuel price *level* alone
explains virtually none of the variance in Transport CPI (R² ≈ 0.0), and the coefficient on fuel price
is statistically insignificant (p = 0.97). **This is a genuine and important finding for logistics
businesses**: budgeting transport/logistics costs purely off the pump price is unreliable in Rwanda;
regulatory, fare-policy and exchange-rate factors matter more for year-level cost levels. The YoY-change
regression below is the more decision-useful model.

```python
# Predict YoY %% change in CPI Transport from YoY %% change in fuel price
Xy = sub[['Fuel_YoY_pct']].values
yy = sub['CPI_YoY_pct'].values
lr_yoy = LinearRegression().fit(Xy, yy)
pred_yoy = lr_yoy.predict(Xy)
print(f"CPI_YoY_% = {lr_yoy.intercept_:.2f} + {lr_yoy.coef_[0]:.2f} * Fuel_YoY_%")
print(f"R-squared = {r2_score(yy, pred_yoy):.4f}")
```

**Interpretation:** For every 1 percentage-point increase in the year-on-year fuel price change,
Transport CPI growth rises by roughly 0.4–0.5 percentage points on average, though the model explains
only ~23% of the variance (R²≈0.23) , plenty of other factors (fare policy, vehicle costs) still drive
the rest. This partial pass-through elasticity is nonetheless a useful planning benchmark for
logistics-cost budgeting.

## 9. Forecasting

```python
# Linear trendline forecast to 2028 for both series (df_cc, 2009-2025)
Xyear = df_cc[['YEAR']].values
future_years = np.array([[2026],[2027],[2028]])

lr_fuel = LinearRegression().fit(Xyear, df_cc['Gasoline_USD_per_L'].values)
fc_fuel = lr_fuel.predict(future_years)
r2_fuel_trend = r2_score(df_cc['Gasoline_USD_per_L'].values, lr_fuel.predict(Xyear))

lr_cpi = LinearRegression().fit(Xyear, df_cc['CPI_Transport_Annual'].values)
fc_cpi = lr_cpi.predict(future_years)
r2_cpi_trend = r2_score(df_cc['CPI_Transport_Annual'].values, lr_cpi.predict(Xyear))

print("Fuel price forecast 2026-2028 (USD/L):", np.round(fc_fuel,3), f"  [trend R2={r2_fuel_trend:.2f}]")
print("CPI Transport forecast 2026-2028:      ", np.round(fc_cpi,2), f"  [trend R2={r2_cpi_trend:.2f}]")
```

```python
fig, axes = plt.subplots(1,2, figsize=(13,5))
all_years = np.concatenate([Xyear.flatten(), future_years.flatten()]).reshape(-1,1)

axes[0].plot(df_cc.YEAR, df_cc.Gasoline_USD_per_L, 'o-', color='#1f4e79', label='Actual')
axes[0].plot(all_years, lr_fuel.predict(all_years), '--', color='#e07b39', label='Linear trend')
axes[0].scatter(future_years, fc_fuel, color='red', zorder=5, label='Forecast')
axes[0].set_title('Fuel Price Forecast', fontweight='bold'); axes[0].legend()

axes[1].plot(df_cc.YEAR, df_cc.CPI_Transport_Annual, 'o-', color='#8b1e3f', label='Actual')
axes[1].plot(all_years, lr_cpi.predict(all_years), '--', color='#c9a227', label='Linear trend')
axes[1].scatter(future_years, fc_cpi, color='red', zorder=5, label='Forecast')
axes[1].set_title('Transport CPI Forecast', fontweight='bold'); axes[1].legend()
plt.tight_layout(); plt.savefig('fig5_fuel_forecast.png'); plt.show()
```

**Interpretation:** A simple linear trendline is a weak fit for fuel price (R² ≈ 0.02 over
2009–2025) because the true pattern is cyclical (boom → bust → flat → shock), not a straight line , 
the model is shown here as the assignment-required "trendline" method, but a logistics planner should
treat the flat forecast (~USD 1.25–1.28/L) with caution given the historical volatility. The Transport
CPI trendline fits far better (R² ≈ 0.87) and projects continued growth to roughly **160–170 index
points by 2028**, i.e. transport/logistics costs are expected to keep rising even if fuel prices
stabilise, reinforcing the earlier finding that non-fuel cost drivers dominate.

## 10. Scenario Analysis :  Best Case vs Worst Case

```python
last_fuel = df_cc.Gasoline_USD_per_L.iloc[-1]   # 2025 pump price = anchor

def scenario_cpi(fuel_change_pct):
    """Predict CPI Transport index using the level-regression, applying a fuel-price shock."""
    new_fuel = last_fuel * (1 + fuel_change_pct/100)
    predicted_cpi = lr.intercept_ + lr.coef_[0] * new_fuel
    return new_fuel, predicted_cpi

scenarios = {
    'Best case (-15% fuel price)':  scenario_cpi(-15),
    'Base case (0% change)':        scenario_cpi(0),
    'Worst case (+25% fuel price)': scenario_cpi(25),
}
for name, (fuel, cpi) in scenarios.items():
    print(f"{name:35s} -> fuel = USD {fuel:.3f}/L | predicted CPI Transport = {cpi:.2f}")
```

```python
fig, ax = plt.subplots(figsize=(8,5))
labels = list(scenarios.keys())
vals = [v[1] for v in scenarios.values()]
colors = ['#2e7d32','#1f4e79','#b71c1c']
ax.bar(labels, vals, color=colors)
for i,v in enumerate(vals):
    ax.text(i, v+1, f'{v:.1f}', ha='center', fontweight='bold')
ax.set_title('Scenario Analysis: Predicted Transport CPI under Fuel-Price Shocks', fontweight='bold')
ax.set_ylabel('Predicted CPI Transport Index')
plt.xticks(rotation=10)
plt.tight_layout(); plt.savefig('fig7_scenario.png'); plt.show()
```

**Interpretation:** Because the level-regression's slope is small and statistically insignificant,
the *level* model predicts only a small direct CPI move even under a severe ±15%/+25% fuel shock ,  again
underlining that **level pass-through is weak**. Using the more realistic **YoY elasticity model**
(≈0.4–0.5), a +25% fuel-price shock would be expected to add roughly **10–12 percentage points** to
transport-cost inflation in that year , a materially larger and more decision-relevant effect. Both
models are reported for transparency; logistics businesses should plan around the YoY-elasticity
scenario as the more realistic one.

```python
# Scenario using the YoY elasticity model (more realistic short-run pass-through)
def scenario_cpi_yoy(fuel_change_pct):
    predicted_cpi_yoy = lr_yoy.intercept_ + lr_yoy.coef_[0] * fuel_change_pct
    return predicted_cpi_yoy

for name, shock in [('Best case (fuel -15%% YoY)', -15), ('Base case (fuel 0%% YoY)', 0), ('Worst case (fuel +25%% YoY)', 25)]:
    print(f"{name:32s} -> predicted CPI Transport YoY change = {scenario_cpi_yoy(shock):.2f}%")
```

## 11. Business Decision-Making : Recommendations

1. **Do not budget logistics costs off the pump price alone.** The level-correlation between fuel
   price and Transport CPI is statistically insignificant , regulatory fare-setting, vehicle costs, and
   FX movements are stronger drivers of year-level transport cost. Logistics firms should track NISR's
   monthly Transport CPI release directly rather than proxying it with fuel prices.
2. **Use the YoY pass-through elasticity (~0.4–0.5) for shock planning.** When global oil prices spike
   and RURA passes part of it through to the pump, firms should expect roughly 40–50% of that percentage
   fuel increase to show up in transport-cost inflation within the same year ,  useful for building
   contingency margins into freight/logistics contracts.
3. **Watch policy events, not just oil markets.** The sharpest CPI-transport jump in the whole series
   (2024) was driven by the government's removal of transport-fare subsidies, not by fuel prices , a
   reminder that regulatory risk is a first-order cost driver in Rwanda's transport sector.
4. **Hedge against RWF depreciation.** Since Rwanda imports 100% of its refined fuel, a weakening
   Rwandan Franc mechanically raises USD-denominated import costs even if world oil prices are flat;
   businesses reliant on imported inputs/transport should monitor BNR exchange-rate policy alongside
   fuel prices.
5. **Build scenario buffers, not point forecasts.** Given the trendline's weak fit for fuel price
   (R²≈0.02), firms should budget a *range* (e.g. the best-/worst-case scenarios above) rather than a
   single-point forecast for 2026–2028 transport costs.


## 12. Conclusion

Rwanda's transport and logistics costs, as captured by the NISR Transport CPI, have risen steadily and
significantly over 2009–2025 ,  more than doubling ,  even though pump fuel prices have moved in cycles
and were administratively held flat for long stretches (2017–2021). The weak level-relationship but
moderate YoY pass-through between fuel prices and transport CPI shows that **fuel price is only one of
several transport-cost drivers** in Rwanda's regulated market; regulatory decisions (fare caps, subsidy
policy) and currency movements are at least as important. Businesses in retail, agribusiness, and import/
export logistics should therefore build cost models that separately track fuel-price pass-through risk,
regulatory/fare-policy risk, and FX risk, rather than treating pump price as a single sufficient
predictor of logistics cost inflation.
