```python
from sklearn.linear_model import LinearRegression
```

## Step 1: Fit the trend

```python
X = df[['month']]
y = df[['sales']]
model = LinearRegression().fit(X,y)
df['trend'] = model.predict(X)
```

## Step 2: Add uncertainty

```python
import pymc as pm
with pm.Model():
    slope = pm.Normal('slope', 569.38, 95.11)
    intercept = pm.Normal('intercept',569.38, 95.11)
    sigma = pm.HalfNormal('sigma', 10)
    mu = intercept + slope * X
    pm.Normal('obs', mu, sigma, observed=y)
    post = pm.sample(1000)
```

## Step 3: Interpret the range

```python
fc = post.posterior
pred = fc ['intercept'] + fc ['slope'] * 19
low, high = pred.quantile([0.1, 0.9])
print(f'{low:.0f} to {high: .0f} units')
```
